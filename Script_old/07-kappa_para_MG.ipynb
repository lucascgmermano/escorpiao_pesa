```python
import pandas as pd
import numpy as np
from sklearn.metrics import cohen_kappa_score
```

```python
dfInd = pd.read_csv('C:/Users/DRS14-SJBV/Downloads/Projetos programacao/Escorpiao_pesa/Quadro2/dados/sinan_mg2.csv', sep=';', encoding='iso-8859-1')
```

```python
# 1. As condições precisam estar dentro de uma lista []
condicoes = [
    (dfInd['TRA_CLASSI'] == 'Moderado') | (dfInd['TRA_CLASSI'] == 'Grave')
]

# 2. As escolhas TAMBÉM precisam ser uma lista [] com o mesmo tamanho
escolha = [1]

# 3. Chame os argumentos na ordem correta: np.select(condlist, choicelist, default)
dfInd['MG_ORIGINAL'] = np.select(condlist=condicoes, choicelist=escolha, default=0)
```

```python
pd.crosstab(dfInd['MG'],dfInd['MG_ORIGINAL'])
```

```python
from sklearn.metrics import cohen_kappa_score

kappa = cohen_kappa_score(
    dfInd["MG_ORIGINAL"],
    dfInd["MG"]
)

print(f"Kappa de Cohen: {kappa:.3f}")
```

```python
a = ((dfInd["MG"] == 1) & (dfInd["MG_ORIGINAL"] == 1)).sum()
b = ((dfInd["MG"] == 1) & (dfInd["MG_ORIGINAL"] == 0)).sum()
c = ((dfInd["MG"] == 0) & (dfInd["MG_ORIGINAL"] == 1)).sum()
d = ((dfInd["MG"] == 0) & (dfInd["MG_ORIGINAL"] == 0)).sum()

concordancia_positiva = (2 * a) / (2 * a + b + c)
concordancia_negativa = (2 * d) / (2 * d + b + c)

print(f"Concordância positiva: {concordancia_positiva:.1%}")
print(f"Concordância negativa: {concordancia_negativa:.1%}")
```

```python
import numpy as np

dfInd['CONCORDANCIA_MG'] = np.select(
    [
        (dfInd['MG'] == 0) & (dfInd['MG_ORIGINAL'] == 0),
        (dfInd['MG'] == 1) & (dfInd['MG_ORIGINAL'] == 1),
        (dfInd['MG'] == 1) & (dfInd['MG_ORIGINAL'] == 0),
        (dfInd['MG'] == 0) & (dfInd['MG_ORIGINAL'] == 1)
    ],
    [
        'Não-MG em ambos',
        'MG em ambos',
        'MG apenas sintético',
        'MG apenas original'
    ],
    default='Indefinido'
)

print(dfInd['CONCORDANCIA_MG'].value_counts())
```

```python
import pandas as pd
import numpy as np
```

```python
sinan_agregado = pd.read_excel('Dados-iniciais/sinan_agregado2.xlsx')
base02 = pd.read_excel('Dados-iniciais/base02.xlsx')
```

```python
# Merge Base02 e Sinan_agregado
base03 = base02.merge(
    on='IBGE',
    how='left',
    right=sinan_agregado
)
```

```python
# Preenche com Zero os valores ausentes
base03 = base03.fillna(0)

# Exclui colunas inuteis
base03 = base03.drop(columns=['Unnamed: 0_x','_merge','merge_flag','Unnamed: 0_y','MUNI_NOME_y'])
```

```python
# Salvar base03
base03.to_excel('Dados-iniciais/base03.xlsx')
```

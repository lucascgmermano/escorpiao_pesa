```python
import pandas as pd
import openpyxl
from unidecode import unidecode
```

```python
df1 = pd.read_excel('Dados-iniciais/Referencia-pesa-municipio-v4.xlsx')
df2 = pd.read_excel('Dados-iniciais/Pesa-enderecos-ms-v3.xlsx')
```

```python
# Normalizando texto df1

colunas_df1 = ['REGIAO', 'PESA_MUNI', 'REFERENCIADO', 'PUBLICO', 'OBSERVACAO', 'muni_limpo']

for i in colunas_df1:
    df1[i] = (
        df1[i]
        .fillna('')              # Substitui valores nulos por texto vazio
        .astype(str)             # Garante que tudo vire texto
        .apply(unidecode)        # Remove acentos
        .str.upper()             # Coloca em caixa alta
    )
```

```python
# Normalizando texto df2

colunas_df2 = ['PESA_MUNI', 'NOME_PESA', 'ENDERECO','ENDERECO_COMPLETO']

for i in colunas_df2:
    df2[i] = (
        df2[i]
        .fillna('')
        .astype(str)
        .apply(unidecode)
        .str.upper()
    )
```

```python
#df1.loc[df1['MULTIPLO_PESA']==0]
df1
```

```python
# Join Pesa + Muni (somente Pesa unico e sem observações)
df3 = df1.loc[(
    (df1['MULTIPLO_PESA']==0) & 
    (df1['PUBLICO']=='TODOS') &
    (df1['OBSERVACAO']=='')
    )].merge(
    right=df2.loc[df2['MULTIPLO_PESA']==0],
    how='left',
    left_on='PESA_MUNI',
    right_on='PESA_MUNI'
)
```

```python
# Join Pesa + Muni Sit. Especiais (Multiplos Pesa e observaçoes)
df4 = df1.loc[(
    (df1['MULTIPLO_PESA']==1) | 
    (df1['OBSERVACAO']!='')
    )].merge(
    right=df2.loc[df2['MULTIPLO_PESA']==1],
    how='left',
    left_on='PESA_MUNI',
    right_on='PESA_MUNI'
)
df4
```

```python
# Salvar df3 e df4
df3.to_excel('Dados-iniciais/juntos_unico_pesa.xlsx')
df4.to_excel('Dados-iniciais/juntos_multi_pesa.xlsx')
```
